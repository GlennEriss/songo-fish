```python
from pathlib import Path
import os, subprocess, sys

REPO_URL = 'https://github.com/GlennEriss/songo-fish.git'
REPO_DIR = Path('/content/songo-fish')
if not REPO_DIR.exists():
    subprocess.run(['git', 'clone', REPO_URL, str(REPO_DIR)], check=True)
else:
    subprocess.run(['git', '-C', str(REPO_DIR), 'pull', '--ff-only'], check=True)
os.chdir(REPO_DIR)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-e', '.[train,perf,dev]'], check=True)
os.environ['PYTHONPATH'] = 'packages:apps/trainer/scripts'
print('Projet installé depuis :', Path.cwd())
```

```python
from google.colab import drive, files
DRIVE_AVAILABLE = False
try:
    drive.mount('/content/drive')
    DRIVE_AVAILABLE = True
except Exception as exc:
    print('Drive indisponible:', exc)

DRIVE_ROOT = Path('/content/drive/MyDrive/songo-ai')
SOURCE = DRIVE_ROOT / 'experiments/lot41_deep_mcts_convergence' if DRIVE_AVAILABLE else Path('/content/lot41_deep_mcts_convergence')
OUTPUT = DRIVE_ROOT / 'experiments/lot41b_deep_mcts_strategic_convergence' if DRIVE_AVAILABLE else Path('/content/lot41b_deep_mcts_strategic_convergence')
EXPORT = DRIVE_ROOT / 'exports/lot41b_results.tar.gz' if DRIVE_AVAILABLE else Path('/content/lot41b_results.tar.gz')
OUTPUT.mkdir(parents=True, exist_ok=True)
EXPORT.parent.mkdir(parents=True, exist_ok=True)
print('SOURCE =', SOURCE)
print('OUTPUT =', OUTPUT)
```

```python
required = [SOURCE / 'search' / f'budget_{b}.json' for b in [256,512,1024,2048,4096,8192,16384,32768]]
for path in required:
    print(path.name, 'OK' if path.exists() else 'MISSING')
assert all(path.exists() for path in required), 'Artefacts Lot41 incomplets; ne pas relancer MCTS ici.'
```

```python
subprocess.run([sys.executable, 'apps/trainer/scripts/analyze_srn_lot41b.py', '--source', str(SOURCE), '--output', str(OUTPUT)], check=True)
subprocess.run([sys.executable, 'apps/trainer/scripts/analyze_srn_lot41b.py', '--stage', 'export', '--output', str(OUTPUT), '--bundle', str(EXPORT)], check=True)
print('decision.json:', OUTPUT / 'decision.json')
print('export:', EXPORT)
if not DRIVE_AVAILABLE:
    files.download(str(EXPORT))
    files.download(str(Path(str(EXPORT) + '.sha256'))
```
